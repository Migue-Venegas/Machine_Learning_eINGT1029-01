# 1. Frameworks
El desarrollo de un proyecto de analítica de datos requiere una estructura que permita ordenar las actividades desde la comprensión inicial del problema hasta la preparación, análisis y evaluación de los datos. En este apartado se revisan tres enfoques ampliamente utilizados: CRISP-DM, KDD y SEMMA, comparando sus etapas principales, alcances y limitaciones.
Esta revisión permite contar con una base metodológica para seleccionar el enfoque más adecuado al contexto del conjunto de datos Store Sales-Time Series Forecasting y al alcance definido para este informe.


## 1.1 CRISP-DM

**CRISP-DM** organiza un proyecto de analítica en seis etapas:

1. **Comprensión del negocio:** identificar el problema, los objetivos y las necesidades del proyecto.
2. **Comprensión de los datos:** conocer las fuentes disponibles, explorar su contenido e identificar problemas iniciales de calidad.
3. **Preparación de los datos:** seleccionar, limpiar, integrar y transformar la información que será utilizada.
4. **Modelado:** aplicar técnicas analíticas o modelos sobre los datos preparados.
5. **Evaluación:** revisar si los resultados obtenidos responden adecuadamente al objetivo planteado.
6. **Despliegue:** organizar la forma en que los resultados o soluciones serán utilizados.

### Alcances
CRISP-DM permite abordar un proyecto de manera integral, considerando tanto el problema que se busca resolver como las etapas técnicas relacionadas con los datos y su análisis. Además, sus etapas pueden revisarse de manera iterativa cuando los resultados de una fase obligan a volver a una etapa anterior.

### Limitaciones
Su estructura es general y no define herramientas o técnicas específicas para cada etapa. Por este motivo, las decisiones relacionadas con limpieza, transformación, modelamiento o implementación deben ser definidas según las características de cada proyecto.

## 1.2 KDD

**KDD** corresponde a un proceso orientado a obtener conocimiento útil a partir de grandes conjuntos de datos. Sus principales etapas pueden resumirse en:

1. **Selección:** definir los datos relevantes para el problema.
2. **Preprocesamiento:** limpiar los datos y corregir problemas que puedan afectar el análisis.
3. **Transformación:** preparar las variables y representar los datos de una manera adecuada para el análisis.
4. **Data Mining:** aplicar métodos que permitan identificar patrones o relaciones dentro de los datos.
5. **Interpretación y evaluación:** analizar los patrones encontrados y determinar si representan información útil.

### Alcances
KDD entrega una estructura especialmente útil cuando el objetivo principal es descubrir patrones y conocimiento a partir de datos existentes. Da una alta importancia a la selección, preparación y transformación de la información antes de aplicar técnicas analíticas.

### Limitaciones
Su enfoque está principalmente orientado al descubrimiento de conocimiento desde los datos, por lo que aspectos como la definición inicial del problema de negocio o la implementación posterior de los resultados tienen menor presencia que en metodologías más amplias.

## 1.3 SEMMA

**SEMMA** es una metodología desarrollada por SAS cuyo nombre corresponde a las etapas **Sample, Explore, Modify, Model y Assess**.

1. **Sample:** seleccionar una muestra de datos que permita realizar el análisis de forma eficiente.
2. **Explore:** estudiar los datos mediante estadísticas y herramientas gráficas para identificar relaciones, tendencias o anomalías.
3. **Modify:** preparar los datos mediante selección, creación o transformación de variables, además del tratamiento de problemas de calidad.
4. **Model:** aplicar diferentes técnicas de modelamiento.
5. **Assess:** evaluar y comparar los resultados obtenidos.

### Alcances
SEMMA entrega una secuencia clara para explorar, preparar y modelar datos. Su etapa de modificación considera tareas directamente relacionadas con este trabajo, como transformación de variables, identificación de outliers y tratamiento de valores faltantes.

### Limitaciones
El método está fuertemente orientado al proceso técnico de análisis y modelamiento de datos. En comparación con CRISP-DM, entrega menos énfasis explícito a la comprensión inicial del problema y a la utilización o despliegue posterior de los resultados.

## 1.4 Comparación de los frameworks

| Característica | CRISP-DM | KDD | SEMMA |
|---|---|---|---|
| Enfoque principal | Desarrollo integral del proyecto analítico | Descubrimiento de conocimiento desde los datos | Análisis y modelamiento de datos |
| Inicio del proceso | Comprensión del problema | Selección de los datos | Muestreo de los datos |
| Exploración de datos | Sí | Sí | Sí |
| Preparación y limpieza | Sí | Sí | Sí |
| Modelamiento | Sí | Data Mining | Sí |
| Evaluación | Sí | Interpretación y evaluación | Sí |
| Considera implementación/despliegue | Sí | Limitado | Limitado |
| Principal fortaleza | Integra problema, datos y resultados | Fuerte énfasis en descubrimiento de patrones | Flujo técnico simple y ordenado |
| Principal limitación | Es general y requiere definir las técnicas específicas | Menor orientación al problema de negocio | Menor énfasis en negocio y despliegue |

## 2.10 Identificación y tratamiento de valores faltantes (Limpieza)

Los valores faltantes pueden afectar la calidad del análisis y, por lo mismo, deben identificarse antes de decidir cómo tratarlos. En esta sección se revisa la cantidad y proporción de datos ausentes en las bases disponibles, con el objetivo de aplicar una solución coherente con la naturaleza de cada variable.

```python
# Identificación de valores faltantes

faltantes = []

for nombre, df in bases.items():

    for columna in df.columns:

        cantidad = df[columna].isnull().sum()

        if cantidad > 0:
            faltantes.append({
                "Base": nombre,
                "Variable": columna,
                "Cantidad": cantidad,
                "Porcentaje (%)": round(
                    cantidad / len(df) * 100, 2
                )
            })

faltantes = pd.DataFrame(faltantes)

faltantes
```

La revisión muestra que el único problema de datos faltantes se concentra en la base `oil`. En particular, la variable `dcoilwtico` presenta 43 valores ausentes, equivalentes aproximadamente al 3,53% de sus registros.
Como se trata de una variable organizada en el tiempo, no resulta recomendable eliminar directamente estas observaciones. Antes de definir el tratamiento, se revisa la ubicación de los datos faltantes dentro de la serie para seleccionar una alternativa que mantenga la continuidad temporal.


### 2.10.1 Ubicación de los valores faltantes

```python
# Fechas donde no existe información del precio del petróleo
oil[oil["dcoilwtico"].isnull()].head(10)

# Visualización inicial de la serie
plt.figure(figsize=(12, 4))

plt.plot(
    oil["date"],
    oil["dcoilwtico"]
)

plt.title("Evolución del precio del petróleo")
plt.xlabel("Fecha")
plt.ylabel("Precio del petróleo")
plt.tight_layout()

plt.show()
```

Los valores faltantes se ubican dentro de una serie temporal que cuenta con registros conocidos antes y después de varios períodos sin información. Esto permite estimarlos utilizando la continuidad natural de la variable a lo largo del tiempo.
Por este motivo, se selecciona la interpolación lineal como método de tratamiento. Esta técnica permite completar cada valor faltante a partir de los datos cercanos en el tiempo, evitando eliminar registros y conservando una trayectoria coherente de la serie.

### 2.10.2 Tratamiento de los datos faltantes mediante interpolación

```python
# Crear una copia de la base antes de realizar modificaciones
oil_limpio = oil.copy()

# Ordenar cronológicamente
oil_limpio = oil_limpio.sort_values("date")

# Identificar registros que originalmente tenían valores faltantes
oil_limpio["dato_imputado"] = oil_limpio["dcoilwtico"].isnull()

# Utilizar la fecha como índice temporal para interpolar
oil_limpio = oil_limpio.set_index("date")

oil_limpio["dcoilwtico"] = (
    oil_limpio["dcoilwtico"]
    .interpolate(method="time", limit_direction="both")
)

# Recuperar date como columna
oil_limpio = oil_limpio.reset_index()

# Verificación
print(
    "Valores faltantes antes:",
    oil["dcoilwtico"].isnull().sum()
)

print(
    "Valores faltantes después:",
    oil_limpio["dcoilwtico"].isnull().sum()
)
```

Los valores faltantes fueron tratados mediante interpolación temporal, utilizando la información disponible en fechas cercanas para estimar los datos ausentes de forma consistente con la evolución de la serie.
Después del tratamiento, la variable `dcoilwtico` no presenta valores faltantes. Además, se incorporó la variable `dato_imputado`, lo que permite identificar qué observaciones fueron completadas y mantener trazabilidad del proceso de limpieza.


## 2.11 Identificación y tratamiento de datos atípicos

Los datos atípicos corresponden a observaciones que se alejan del comportamiento general de una variable. Sin embargo, no todo valor extremo debe interpretarse automáticamente como un error, ya que en algunos casos puede reflejar situaciones reales del negocio o particularidades de ciertos productos.
En este análisis la revisión se concentra principalmente en la variable `sales`, porque corresponde a la variable objetivo del conjunto de entrenamiento y es clave para cualquier etapa posterior de modelamiento.


### 2.11.1 Identificación mediante rango intercuartílico (IQR)

La identificación de valores atípicos mediante el rango intercuartílico (IQR) permite detectar observaciones inusualmente altas o bajas. Para ello, se calculan límites a partir del rango donde se concentra el 50% central de los datos, lo que facilita reconocer valores que se apartan del comportamiento más frecuente.

```python
# Cálculo de límites mediante rango intercuartílico (IQR)

Q1 = train["sales"].quantile(0.25)
Q3 = train["sales"].quantile(0.75)

IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

outliers_sales = (
    (train["sales"] < limite_inferior) |
    (train["sales"] > limite_superior)
)

print(f"Q1: {Q1:.2f}")
print(f"Q3: {Q3:.2f}")
print(f"Límite superior: {limite_superior:.2f}")
print("Registros identificados:", outliers_sales.sum())
print(f"Porcentaje: {outliers_sales.mean() * 100:.2f}%")
```

Al aplicar inicialmente el criterio IQR sobre `sales`, obtuvimos que un 14,9% de los registros era clasificado como atípico.
Este resultado debe analizarse con cautela, ya que el comportamiento de las ventas cambia de manera importante entre familias de productos. Por lo tanto, una venta alta puede ser normal dentro de una categoría, aunque parezca extrema al compararla con la base completa.

### 2.11.2 Revisión considerando la familia de producto

```python
# Crear copia de la base para mantener los datos originales
train_limpio = train.copy()

# Calcular Q1 y Q3 para cada familia de productos
Q1_familia = (
    train
    .groupby("family")["sales"]
    .transform("quantile", q=0.25)
)

Q3_familia = (
    train
    .groupby("family")["sales"]
    .transform("quantile", q=0.75)
)

IQR_familia = Q3_familia - Q1_familia

# Identificar valores extremos considerando cada familia
train_limpio["outlier_sales"] = (
    (train["sales"] < (Q1_familia - 1.5 * IQR_familia)) |
    (train["sales"] > (Q3_familia + 1.5 * IQR_familia))
)

print(
    "Registros identificados como atípicos:",
    train_limpio["outlier_sales"].sum()
)

print(
    f"Porcentaje: "
    f"{train_limpio['outlier_sales'].mean() * 100:.2f}%"
)
```

Al repetir el análisis considerando cada familia de productos, el porcentaje de observaciones identificadas como atípicas disminuye aproximadamente a un 6,7%. Este resultado es más representativo, ya que compara cada registro con productos de comportamiento similar.
**En este trabajo se decide conservar estos registros, porque no existe evidencia suficiente para clasificarlos como errores de captura o digitación. En lugar de eliminarlos, se incorpora la variable `outlier_sales`, que permite identificarlos y mantener trazabilidad sobre estas observaciones**.

### 2.11.3 Verificación de los valores no válidos

No se identificaron valores negativos en `sales` ni en `onpromotion`, por lo que no se observan registros evidentemente inválidos en estas variables.
En el caso de `onpromotion`, no se aplicó eliminación mediante IQR, debido a que la mayoría de sus registros son iguales a cero. Bajo esta condición, el criterio podría marcar observaciones de manera poco representativa y no necesariamente asociadas a errores.

# 3. Ventanas de entrenamiento, validación y test

La división de los datos en conjuntos de entrenamiento, validación y prueba permite separar las etapas de aprendizaje, ajuste y evaluación. El conjunto de entrenamiento se utiliza para ajustar un eventual modelo, validación permite comparar configuraciones y test se reserva para evaluar finalmente el desempeño sobre datos no utilizados en las etapas anteriores.
Debido a que el conjunto Store Sales presenta una estructura temporal, la separación debe respetar el orden cronológico de las observaciones, evitando que información futura sea utilizada para analizar períodos anteriores.

## 3.1 Métodos de separación considerados

| Método | Descripción | Ventaja | Limitación |
|---|---|---|---|
| **Partición fija Train/Validation/Test** | Se divide la base en tres subconjuntos con funciones diferenciadas. | Es simple de aplicar e interpretar. | El resultado puede depender de los períodos o registros seleccionados. |
| **Validación cruzada (Cross-Validation)** | Se realizan distintas divisiones internas de los datos para evaluar el desempeño de forma más robusta. | Reduce la dependencia de una única partición. | En su forma tradicional puede no respetar el orden temporal. |
| **Ventanas temporales** | Se utilizan datos históricos para entrenamiento, un período posterior para validación y uno más reciente para prueba. | Respeta la secuencia temporal y representa mejor una situación de predicción futura. | La evaluación depende de la selección de las ventanas de tiempo. |

Para este trabajo se selecciona el uso de **ventanas temporales**, debido a que los registros corresponden a una serie ordenada cronológicamente.

## 3.2 Criterio utilizado para definir las ventanas

Para aplicar la separación temporal se definieron períodos calendario fáciles de identificar e interpretar:

- **Entrenamiento:** enero de 2013 a diciembre de 2015.
- **Validación:** enero a diciembre de 2016.
- **Test:** enero de 2017 al 15 de agosto de 2017.

Esta selección corresponde a una decisión metodológica del presente trabajo.  Su objetivo es mantener períodos cronológicos claramente diferenciados.

## 3.3 Conjuntos de entrenamiento, validación y test

Los datos se dividirán en tres conjuntos:

- **Training:** utilizado para que un eventual modelo aprenda los patrones existentes en los datos.
- **Validation:** utilizado para comparar configuraciones o realizar ajustes sin utilizar todavía el conjunto de prueba.
- **Test:** reservado para evaluar finalmente el comportamiento sobre datos que no participaron en las etapas anteriores.

El conjunto de test debe mantenerse separado durante el desarrollo para evitar que influya en las decisiones tomadas durante el entrenamiento o validación.

## 3.4 Aplicación de la separación temporal

```python
# Ordenar la base cronológicamente
train_limpio = (
    train_limpio
    .sort_values("date")
    .reset_index(drop=True)
)

# Definir ventanas temporales

datos_train = train_limpio[
    (train_limpio["date"] >= "2013-01-01") &
    (train_limpio["date"] <= "2015-12-31")
].copy()

datos_validation = train_limpio[
    (train_limpio["date"] >= "2016-01-01") &
    (train_limpio["date"] <= "2016-12-31")
].copy()

datos_test = train_limpio[
    (train_limpio["date"] >= "2017-01-01") &
    (train_limpio["date"] <= "2017-08-15")
].copy()
```

## 3.5 Verificación de la separación

```python
print("TRAIN")
print("Registros:", len(datos_train))
print("Período:", datos_train["date"].min(), "a", datos_train["date"].max())

print("\nVALIDATION")
print("Registros:", len(datos_validation))
print("Período:", datos_validation["date"].min(), "a", datos_validation["date"].max())

print("\nTEST")
print("Registros:", len(datos_test))
print("Período:", datos_test["date"].min(), "a", datos_test["date"].max())

```

```python
#Comprobación de la división
total_dividido = (
    len(datos_train) +
    len(datos_validation) +
    len(datos_test)
)

print("Registros base original:", len(train_limpio))
print("Registros después de dividir:", total_dividido)
```

## 3.6 Resultado de la separación

La base fue dividida en tres ventanas cronológicas. El conjunto de entrenamiento utiliza información entre 2013 y 2015, validación corresponde al año 2016 y test contiene los registros disponibles de 2017 hasta el 15 de agosto.
La separación conserva la totalidad de los registros y mantiene el orden temporal, evitando que información de períodos posteriores forme parte de las etapas anteriores. De esta forma, el conjunto de test representa el período más reciente disponible dentro de la base histórica.
